# Visión del problema y selección de variables

## Definir un problema de Machine Learning realista (clasificación, regresión, clustering, etc.).

El problema consiste en la predicción espacial de la temperatura de superficie terreste(LST, por su siglas en inglés) a nivel intraurbano. El objetivo del modelo es aprender patrones termodinámicos entre la morfología de la ciudad y el calor, actuando no solo como un modelo predictivo estatico, sino como el motor de inferencia que permita simular intervenciones urbanísticas.

## Describir el contexto del problema y su utilidad.

- Contexto: La ciudad de Sevilla sufre episodios de calor extremo durante los meses de verano, agravados drásticamente por el efecto de la Isla de Calor Urbana(UHI por su siglas en ingles). El sellado del suelo con asfalto, la falta de vegetación y la densidad de los edificios, junto con otras variables, provocan que ciertas areas retengan el calor solar, teniendo consecuencias en la salud publica, la habitabilidad y el consumo energetico.

- Utilidad: La utilidad de este modelo se encuentra en permitir a urbanistas, administraciones y ciudadanos realiazar simulaciones "What-if" en tiempo real, al integrar el modelo entrenado en una aplicación web interactiva. Los usuarios podran valorar matematicamente la temperatura termica de un barrio concreto modificando variables como el arbolado, desificacion de edificios o la aplicacion de materiales reflectantes.


## Identificar las variables relevantes que influyen en el problema.


### Justificar la selección de dichas variables.

Usando datos satelitales y datos morfoloficos, se intentaran obtener una serie de variables que a priori tienen impacto directo en la termodinamica urbana.

Variable objetivo: 

- Temperatura (LST): Representa la temperatura de superficie terrestre en gracods Celcius.

Variables independientes:

- NDVI (Indice de vegetacion diferenciada normalizada): La vegetación mitiga el calor mediante la evapotranspiración(las plantas liberan vapor de agua) y la absorción de radiacion para la fotosintesis.
 
- NDBI(Indice de suelo edificado): Cuantifica el nivel de asfalto y hormigon, materiales que absorven el calor diurno y lo liberan durante el resto del día.

- Albedo: Mide la reflectancia de la sufercie, un alto albedo(suerficies claras) reflejan la radiacion solar incidente, evitando que el material se caliente.

- Densidad de arbolado: Mientras que el NDVI mide el verder general(incluyendo cesped), la densidad de arboles captura el efecto de sombreado, que reduce drasticamente el impacto de la radiación directa a pie de calle.

- Densidad de edificios y media de altura: Estas variables pretenden controlas el "Efecto cañon urbano". Una alta densidad y altura de los edificios puede atrapar el calor y bloquear el viento, pero al mismo tiempo proporciona sombras continuas en las calles durante el dia.


> [!NOTE] Es posible que durante las obtenciones de diferentes datos se añadan nuevas variables que se observe que pueden aportar información.

## Tipo de problema (supervisado / no supervisado).

Esto es un problema de aprendzaje Supervisado, donde la variable objetivo se la proporcionamos al modelo.
